## 4) Data collection and storage
Enter the measurements done: 1) diameter of spheres, 2) mass of spheres, 3) volume of fluid, 4) masss of fluid, 5) sedimentation time, 6) sedimentation distance

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Enter measured sphere size in [mm]
d1 = [5.44,5.42,5.41,5.44,5.42,5.44]
d2 = [5.43,5.43,5.43,5.42,5.44,5.42]
d3 = [5.40,5.43,5.42,5.41,5.42,5.40,5.41]
d4 = [1]
d5 = [2.12,2.09,2.07,2.06]
d6 = [6.00,6.05,5.90,5.89,5.94,6.02,5.98,6.03]
d1 = np.array(d1, dtype=float)
d2 = np.array(d2, dtype=float)
d3 = np.array(d3, dtype=float)
d4 = np.array(d4, dtype=float)
d5 = np.array(d5, dtype=float)
d6 = np.array(d6, dtype=float)

# compute diameter average and change units [mm]-->[cm]
D1 = np.mean(d1)/10
D2 = np.mean(d2)/10
D3 = np.mean(d3)/10
D4 = np.mean(d4)/10
D5 = np.mean(d5)/10
D6 = np.mean(d6)/10
D = [D1,D2,D3,D4,D5,D6]
D = np.array(D, dtype=float)
D = D

# Enter the number of measured spheres (for their mass)
n1 = 15 
n2 = 13
n3 = 19
# spheres of type 4 are made of same material as 5 and 6
n4 = 0
n5 = 50 
n6 = 14

# Enter measured masses for spheres [g]
ms1 = 1.95/n1
ms2 = 1.33/n2
ms3 = 1.9/n3
# spehre of type 3 are not measured as they are too small
ms4 = 0
ms5 = 0.54/n5
ms6 = 4.0/n6

# density in kg per m3
rho1 = ms1 / (4/3*np.pi * (D1/2)**3) * 1000
rho2 = ms2 / (4/3*np.pi * (D2/2)**3) * 1000
rho3 = ms3 / (4/3*np.pi * (D3/2)**3) * 1000
rho5 = ms5 / (4/3*np.pi * (D5/2)**3) * 1000
rho6 = ms6 / (4/3*np.pi * (D6/2)**3) * 1000
rho4 = (rho5 + rho6)/2
rho_s = [rho1,rho2,rho3,rho4,rho5,rho6]
rho_s = np.array(rho_s, dtype=float)

# Enter sedimentation distance in [m]
L = 0.199

# Enter sedimentation time in [s]
t1 = [3.3,3.21,3.28,3.21,3.18,3.18]
t2 = [2.03,2.18,1.95,2.06,1.91,2.10,2.10]
t3 = [0]
t4 = [36.47]
t5 = [10.06,9.89]
t6 = [1.68,1.46,1.88,1.78,1.80]

t1 = np.array(t1, dtype=float)
t2 = np.array(t2, dtype=float)
t3 = np.array(t3, dtype=float)
t4 = np.array(t4, dtype=float)
t5 = np.array(t5, dtype=float)
t6 = np.array(t6, dtype=float)

# compute diameter average
T1 = np.mean(t1)
T2 = np.mean(t2)
T3 = np.mean(t3)
T4 = np.mean(t4)
T5 = np.mean(t5)
T6 = np.mean(t6)
Ts = [T1,T2,T3,T4,T5,T6]

# sedimentation velocity
vs1 = L / t1
vs2 = L / t2
vs3 = L / t3
vs4 = L / t4
vs5 = L / t5
vs6 = L / t6

# Enter volume [cm3] and mass [g] of fluid mesaured
Vf = 1
mf = 1.22

# fluid density
rho_f = mf / Vf * 1000
rho_f = np.array(rho_f, dtype=float)

df = pd.DataFrame({"Diameter, [cm]": D, "Sp. density, [kg/m3]": rho_s, "Sed. velocity, [m/s]": vs, "Fluid density, [kg/m3]": rho_f})
df
# optional: round values for nicer display
df = df.round({
    "Diameter, [cm]": 3,
    "Sp. density, [kg/m3]": 1,
    "Sed. velocity, [m/s]": 3,
    "Fluid density, [kg/m3]": 1
})
df
```

## 5) Show results


```python
x = rho_s - rho_f

plt.figure(1)
plt.plot(D,vs/x,'.')
plt.ylabel(r"Sedimentation velocity, [cm/s]")
plt.xlabel(r"Sphere diameter ($D$), [cm]")
plt.grid(False)
plt.show()

plt.figure(2)
plt.plot(x,vs/D**2,'.')
plt.ylabel(r"vs / D^2, [cm/s]")
plt.xlabel(r"Density difference - $\rho_s - \rho_f$, [g/mL]")
plt.grid(False)
plt.show()
```

## 6) Data analysis

```python
# compute the fluid viscosity using Stokes law
# convert all quantities in kg, m and s
D = D / 100
rho_s1 = rho_s * 1000
rho_f = rho_f * 1000

g = 9.81
mu1 = 2/9 * ((rho1 - rho_f) * (D1/2)**2 * g) / vs1
mu2 = 2/9 * ((rho2 - rho_f) * (D2/2)**2 * g) / vs2
mu3 = 2/9 * ((rho3 - rho_f) * (D3/2)**2 * g) / vs3
mu4 = 2/9 * ((rho4 - rho_f) * (D4/2)**2 * g) / vs4
mu5 = 2/9 * ((rho5 - rho_f) * (D5/2)**2 * g) / vs5
mu6 = 2/9 * ((rho6 - rho_f) * (D6/2)**2 * g) / vs6

# arit. av for each sphere
mu_av1 = np.mean(mu1)
mu_av2 = np.mean(mu2)
mu_av3 = np.mean(mu3)
mu_av4 = np.mean(mu4)
mu_av5 = np.mean(mu5)
mu_av6 = np.mean(mu6)
MU = [mu_av1,mu_av2,mu_av3,mu_av4,mu_av5,mu_av6]

# weights as std
w1 = np.std(mu1)
w2 = np.std(mu2)
w3 = np.std(mu3)
w4 = np.std(mu4)
w5 = np.std(mu5)
w6 = np.std(mu6)
w = w1+w2+w3+w4+w5+w6
W = mu_av1*w1 + mu_av2*w2 + mu_av3*w3 + mu_av4*w4 + mu_av5*w5 + mu_av6*w6
# weigthed average for viscosity
mu = W / w
dmu = 1 / np.sqrt(w)

plt.figure(1)
plt.plot(D,MU,'.')
plt.ylabel(r"viscosity, [Pa s]")
plt.xlabel(r"Sphere diameter ($D$), [cm]")
plt.grid(False)
plt.show()

print(f"Fluid viscosity mu = {mu:.4g} Pa s +- {dmu:.4g} Pa s")
```
